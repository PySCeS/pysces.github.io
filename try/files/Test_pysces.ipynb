## Importing PySCeS and running tests

```python
import numpy as np
import pandas as pd
import pysces
pysces.enablePandas()
```

```python
pysces.test()
```

## Basic functionality using simple test model

```python
mod1 = pysces.model('pysces_test_branch1')
mod1.SetQuiet()
```

### Time simulation

```python
mod1.doSimPlot(points=100, end=20)
```

```python
mod1.SimPlot('rates')
```

Simulation data:

```python
mod1.sim
```

Change `Vf2` (*Vmax* of one of the branch enzymes) and repeat simulation:

```python
mod1.Vf2 = 20
mod1.doSimPlot(points=100, end=20)
```

```python
mod1.SimPlot('rates')
```

### Steady state and MCA

```python
mod1.doStateShow()
```

```python
mod1.doMca()
```

```python
mod1.showCC()
```

### Parameter scans

We are going to scan the value of `Vf2` from 1 to 50 and investigate the effect on the steady state (fluxes and concentrations):

```python
mod1.scan_in = 'Vf2'
mod1.scan_out = ['s1_ss', 's2_ss', 's3_ss', 's4_ss', 'J_R1', 'J_R2', 'J_R3']
scan_range = np.linspace(1, 50, 50)
```

```python
mod1.Scan1(scan_range)
```

```python
mod1.Scan1Plot(['s1_ss', 's2_ss', 's3_ss', 's4_ss'])
```

```python
mod1.Scan1Plot(['J_R1', 'J_R2', 'J_R3'])
```

Scan data:

```python
mod1.scan
```

## Realistic model: sucrose accumulation in sugarcane

This model is described in [Rohwer and Botha (2001) Biochem. J.](https://pmc.ncbi.nlm.nih.gov/articles/PMC1222077/)

```python
mod2 = pysces.model('rohwer_sucrose2', dir='models')
mod2.SetQuiet()
```

### Steady state

```python
mod2.doStateShow()
```

### MCA

```python
mod2.doMca()
```

```python
mod2.showCC()
```

```python
mod2.showEvar()
```

### Parameter scan

```python
mod2.scan_in = 'Vmax9'
mod2.scan_out = ['Suc_ss', 'Fru_ss', 'Glc_ss', 'J_R9', 'J_R11', 'ecR9_Suc',
                'ccJR9_R9', 'ccJR11_R9']
scan_range = np.linspace(0.1, 1.0, 91)
mod2.Scan1(scan_range)
```

Scan data:

```python
mod2.scan
```

Plots:

```python
mod2.Scan1Plot(['Suc_ss', 'Fru_ss', 'Glc_ss'])
mod2.Scan1Plot(['J_R9', 'J_R11'])
mod2.Scan1Plot(['ecR9_Suc', 'ccJR9_R9', 'ccJR11_R9'])
```

## Read in SBML model from BioModels and perform simulation

This is a model of the *Repressilator* by Elowitz (2000), available on [BioModels](https://www.ebi.ac.uk/biomodels/BIOMD0000000012)

```python
mod3 = pysces.loadSBML('BIOMD0000000012_url.xml', sbmldir='models', pscdir='models')
```

### Time simulation and plot

`CVODE` is not available in a web-assembly environment, so we have to use `LSODA`.

```python
mod3.mode_integrator = 'LSODA'
mod3.doSim(end=600, points=300)
mod3.SimPlot(['PX','PY','PZ'])
```

Simulation data:

```python
mod3.sim
```
